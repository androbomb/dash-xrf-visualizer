# XRF visualizer — example

Set `path_to_file` to a `.npz`, `.h5` or `.npy` XRF cube (key `img`, shape (H, W, C)).
The data path is a variable here on purpose — never hardcode it in the app.

```python
#path_to_file = "./data_example/Mantegna_mappa3b_centro-sx_35kV-90uA.npz"  # e.g. "/path/to/Mantegna_mappa3b_centro-sx_35kV-90uA.npz"
path_to_file = "./data_example/Ragazzo-triste_area4_centrodx_38-80_calibrated.npz"  # e.g. "/path/to/Mantegna_mappa3b_centro-sx_35kV-90uA.npz"

from xrf_visualizer.data import load_cube
from xrf_visualizer.ui   import build_app

cube = load_cube(path_to_file).array
print(path_to_file, cube.shape)
app = build_app(cube)
app.run(
    jupyter_mode='inline', 
    debug=False,
    #jupyter_height=800,
)  # or app.run() from a terminal
```
